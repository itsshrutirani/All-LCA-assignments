```python
# ASSIGNMENT NO. 1
# DIFFERENT OPERATIONS ON DICTIONARY AND TUPLE DATA STRUCTURE

#LIST OPERATIONS 
print("LIST OPERATIONS")
student_list = ["shruti", "aarya", "anushka", "akansha", "alfiya"]
print("Original List:", student_list)
student_list.append("palak")
print("After Append:", student_list)
student_list.insert(1, "priya")
print("After Insert:", student_list)
student_list.remove("alfiya")
print("After Remove:", student_list)
student_list.pop()
print("After Pop:", student_list)
student_list.sort()
print("After Sort:", student_list)


# TUPLE OPERATIONS 
print("\nTUPLE OPERATIONS")
student_tuple = ("anandita", "mohit", "divya", "aayush", "sameer")
print("Original Tuple:", student_tuple)
print("Count of mohit:", student_tuple.count("mohit"))
print("Index of divya:", student_tuple.index("divya"))
student_tuple = student_tuple + ("palak",)
print("After Adding Element:", student_tuple)
print("Sliced Tuple:", student_tuple[1:4])
print("Length of Tuple:", len(student_tuple))


# DICTIONARY OPERATIONS 
student_dict = {
    101: "prachi",
    102: "samprikta",
    103: "shshank",
    104: "swanandi",
    105: "shubham"
}
print("\nDICTIONARY OPERATIONS")
print("Original Dictionary:", student_dict)
student_dict[106] = "palak"
print("After Adding Key 106:", student_dict)
student_dict[101] = "priya"
print("After Updating Key 101:", student_dict)
student_dict.pop(102)
print("After Removing Key 102:", student_dict)
print("Keys:", student_dict.keys())
print("Values:", student_dict.values())
print("Items:", student_dict.items())
print("Value using get():", student_dict.get(103))
```
