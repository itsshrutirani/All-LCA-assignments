```python
def Check_right_angles_triangle(a,b,c):
    sides=sorted([a,b,c])
    if sides[0]+sides[1]<=sides[2]:
        return "Triangle cannot be formed!"

    if sides[0]**2+sides[1]**2==sides[2]**2:
            return "Given sides make a right angled triangle"
    else:
        return "Given sides don't make a right angled triangle"

        
x=float(input("Enter first side:"))
y=float(input("Enter second side:"))
z=float(input("Enter third side:"))


print(x)
print(y)
print(z)
print(Check_right_angles_triangle(x,y,z))
```
