```python
import pandas as pd

df = pd.read_csv("C:/Users/admin/OneDrive/Desktop/student_data.csv")

print("Original Dataset:")
print(df)

# Display first five records
print("\nFirst Five Records:")
print(df.head())

# Display dataset information
print("\nDataset Information:")
df.info()

# Display statistical description
print("\nStatistical Description:")
print(df.describe())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Fill missing marks with median
df['Marks'] = df['Marks'].fillna(df['Marks'].median())

# Remove duplicate records
df = df.drop_duplicates()

# Remove irrelevant column
df = df.drop(columns=['Age'])

# Filter students scoring above 80
print("\nStudents Scoring Above 80:")
print(df.loc[df['Marks'] > 80])

# Display final cleaned dataset
print("\nFinal Cleaned Dataset:")
print(df)
```
